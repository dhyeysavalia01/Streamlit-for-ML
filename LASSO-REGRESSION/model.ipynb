```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso, LinearRegression
```

```python
df = pd.read_csv('calories.csv')
df.head()
```

```python
# features and target
x = df.iloc[:, 0:-1]
y = df.iloc[:, -1]

# split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

## Linear Regression

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_Pred = lr.predict(x_test)

print('intercept: ',lr.intercept_)
print('coef: ',lr.coef_)
print('r2_score: ',r2_score(y_test, y_Pred))
```

```python
# plane of Regression
x_range = np.linspace(df['age'].min(), df['age'].max(), 15)
y_range = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(), 15)

xx,yy = np.meshgrid(x_range, y_range)
heart_mean = df['heart_rate'].mean()
heart_col = np.full(xx.ravel().shape, heart_mean)

pred_input = np.c_[xx.ravel(), yy.ravel(), heart_col]
zz = lr.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, x = df['age'],
                    y = df['exercise_min'],
                    z = df['calories_burned'],
                    color=df['calories_burned'],
                    color_continuous_scale='thermal')
fig.add_surface(x = x_range, y = y_range, z = zz, colorscale='thermal',showscale=False)
fig.show()
```

## Lasso Regression

```python
l = Lasso(alpha = 0.1)
l.fit(x_train, y_train)
y_pred = l.predict(x_test)

print('intercept: ',l.intercept_)
print('coef: ',l.coef_)
print('r2_score: ',r2_score(y_test, y_pred))
```

```python
# plane of Regression
x_range = np.linspace(df['age'].min(), df['age'].max(), 15)
y_range = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(), 15)

xx,yy = np.meshgrid(x_range, y_range)
heart_mean = df['heart_rate'].mean()
heart_col = np.full(xx.ravel().shape, heart_mean)

pred_input = np.c_[xx.ravel(), yy.ravel(), heart_col]
zz = l.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, x = df['age'],
                    y = df['exercise_min'],
                    z = df['calories_burned'],
                    color=df['calories_burned'],
                    color_continuous_scale='magma')
fig.add_surface(x = x_range, y = y_range, z = zz, colorscale='magma',showscale=False)
fig.show()
```
