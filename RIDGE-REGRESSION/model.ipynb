```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import plotly.express as px
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
```

```python
df = pd.read_csv('advertising.csv')
df.head()
```

```python
# features and target
x = df.iloc[:, 0:-1]
y = df.iloc[:, -1]

# train test split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
df.corr()
```

```python
# true data
fig = px.scatter_3d(df, 
                    x = df['TV'], 
                    y = df['Radio'], 
                    z = df['Sales'], 
                    color = df['Sales'], 
                    color_continuous_scale='thermal')
fig.show()
```

## 🔰 Linear Regression

```python
# linear Regression
lr = LinearRegression()
lr.fit(x_train, y_train)
y_Pred = lr.predict(x_test)


print('intercept: ',lr.intercept_)
print('coef: ',lr.coef_)
print('R2 score: ',r2_score(y_test, y_Pred))
```

```python
# plane of regression
x_range = np.linspace(df['TV'].min(), df['TV'].max(), 15)
y_range = np.linspace(df['Radio'].min(), df['Radio'].max(), 15)

xx,yy = np.meshgrid(x_range, y_range)
news_mean = df['Newspaper'].mean()
news_col = np.full(xx.ravel().shape, news_mean)

pred_input = np.c_[xx.ravel(), yy.ravel(), news_col]
zz = lr.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, 
                    x = df['TV'], 
                    y = df['Radio'], 
                    z = df['Sales'], 
                    color = df['Sales'], 
                    color_continuous_scale='thermal')

fig.add_surface(x = x_range, y = y_range, z = zz, colorscale='twilight')
fig.show()
```

## 🚀 Ridge Regression

```python
r = Ridge(alpha = 0.001)
r.fit(x_train, y_train)
y_pred = r.predict(x_test)

print('Intercept: ',r.intercept_)
print('Coef: ',r.coef_)
print('R2 Score: ',r2_score(y_test, y_pred))
```

```python
x_range = np.linspace(df['TV'].min(), df['TV'].max(), 15)
y_range = np.linspace(df['Radio'].min(), df['Radio'].max(), 15)

xx,yy = np.meshgrid(x_range, y_range)
news_mean = df['Newspaper'].mean()
news_col = np.full(xx.ravel().shape, news_mean)

pred_input = np.c_[xx.ravel(), yy.ravel(), news_col]
zz = r.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, x = df['TV'],
                    y = df['Radio'],
                    z = df['Sales'],
                    color=df['Sales'],
                    color_continuous_scale='twilight',
                    title='True data')
fig.add_surface(x = x_range, y = y_range, z = zz, colorscale='magma')

fig.show()
```
